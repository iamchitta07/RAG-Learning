## Word Document Processing

```python
from langchain_community.document_loaders import (
    Docx2txtLoader, 
    UnstructuredWordDocumentLoader
)
```

```python
from docx import Document as DocxDocument
import os 
```

```python
# Method 1: Using Docx2txtLoader
print("Loading DOCX using Docx2txtLoader...")
try:
    docx_loader = Docx2txtLoader("data/word_files/proposal.docx")
    docs = docx_loader.load()
    print(f"Loaded {len(docs)} document(s) from the DOCX file.")
    print(docs[0].page_content[:200])  # Print the first 200
    print(f"Metadata: {docs[0].metadata}")
except Exception as e:
    print(f"Error loading DOCX with Docx2txtLoader: {e}")
```

```python
# Method 2: Using UnstructuredWordDocumentLoader
print("\nLoading DOCX using UnstructuredWordDocumentLoader...")
 
try:
    unstructured_loader = UnstructuredWordDocumentLoader("data/word_files/proposal.docx", mode = "elements")
    docs_unstructured = unstructured_loader.load()
    print(f"Loaded {len(docs_unstructured)} document(s) from the DOCX file.")
    for i, doc in enumerate(docs_unstructured[:5]):
        print(f"Document {i+1}")
        print(f"Type: {docs_unstructured[i].metadata.get('category', 'Unknown')}")  # Print the type
        print(f"Content: {docs_unstructured[i].page_content[:200]}")  # Print the first 200 characters
except Exception as e:
    print(f"Error loading DOCX with UnstructuredWordDocumentLoader: {e}") 
```
